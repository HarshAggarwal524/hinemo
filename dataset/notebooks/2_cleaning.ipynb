# HinEmo — Cleaning Pipeline
Reads the merged YouTube data (`data/interim/youtube_merged.csv`) and applies
Step 1.2 cleaning rules: dedup, min length, spam/URL removal. Code-mix ratio
filtering happens in a later notebook once language tags exist (IndicLID or
GPT-based) — this pass handles everything that doesn't need that yet.

```python
import pandas as pd
import re
import os
import string
from collections import Counter

SOURCE_FILES = [
    "../data/raw/youtube_raw.csv",
    "../data/raw/youtube_generalized_raw.csv",
]

frames = []
for path in SOURCE_FILES:
    if os.path.exists(path):
        f = pd.read_csv(path)
        f["_source_file"] = path
        frames.append(f)
        print(f"Loaded {len(f)} rows from {path}")
    else:
        print(f"[!] {path} not found, skipping")

df = pd.concat(frames, ignore_index=True)
before = len(df)
df = df.drop_duplicates(subset=["source_id"], keep="first")
print(f"\nMerged: {before} -> {len(df)} unique rows ({before - len(df)} duplicates removed)")
df.head()
```

```python
print("By target_emotion (blank = generalized/unlabeled):")
print(df["target_emotion"].fillna("(blank)").value_counts())
```

```python
URL_RE = re.compile(r"https?://\S+|www\.\S+")
MENTION_RE = re.compile(r"@\w+")

def basic_clean(text):
    text = URL_RE.sub("", str(text))
    text = MENTION_RE.sub("", text)
    return text.strip()

df["text_clean"] = df["text"].apply(basic_clean)
print(f"Created text_clean for {len(df)} rows")
df[["text", "text_clean"]].head()
```

```python
def token_count(text):
    return len(text.split())

MIN_TOKENS = 5

before_length = len(df)
df = df[df["text_clean"].apply(token_count) >= MIN_TOKENS]
print(f"Before length filter: {before_length}")
print(f"After length filter (>= {MIN_TOKENS} tokens): {len(df)} (-{before_length - len(df)})")
```

```python
df.sample(15)[["text_clean", "target_emotion"]]
```

```python
OTHER_SCRIPT_RANGES = {
    "gurmukhi_punjabi": (0x0A00, 0x0A7F),
    "bengali": (0x0980, 0x09FF),
    "tamil": (0x0B80, 0x0BFF),
    "telugu": (0x0C00, 0x0C7F),
    "kannada": (0x0C80, 0x0CFF),
    "malayalam": (0x0D00, 0x0D7F),
    "gujarati": (0x0A80, 0x0AFF),
    "odia": (0x0B00, 0x0B7F),
}

def detect_other_scripts(text):
    found = set()
    for char in str(text):
        cp = ord(char)
        for script, (start, end) in OTHER_SCRIPT_RANGES.items():
            if start <= cp <= end:
                found.add(script)
    return found

df["other_scripts"] = df["text_clean"].apply(detect_other_scripts)
df["has_other_script"] = df["other_scripts"].apply(lambda s: len(s) > 0)

print(f"Rows with non-Hindi Indic script content: {df['has_other_script'].sum()} "
      f"({df['has_other_script'].sum() / len(df):.1%})")

all_scripts = Counter()
for s in df["other_scripts"]:
    all_scripts.update(s)
print("\nBreakdown by script:")
for script, count in all_scripts.most_common():
    print(f"  {script}: {count}")
```

```python
for script in all_scripts:
    subset = df[df["other_scripts"].apply(lambda s: script in s)]
    print(f"\n=== {script} sample ({len(subset)} total) ===")
    display(subset.sample(min(5, len(subset)))[["text_clean"]])
```

```python
before_script_filter = len(df)
df = df[~df["has_other_script"]].copy()
print(f"Before: {before_script_filter}")
print(f"After dropping other-script rows: {len(df)} (-{before_script_filter - len(df)})")

df = df.drop(columns=["other_scripts", "has_other_script"])

df.to_csv("../data/interim/youtube_cleaned_stage1.csv", index=False)
print(f"\nSaved checkpoint: {len(df)} rows -> data/interim/youtube_cleaned_stage1.csv")
```

```python
import pandas as pd

df_stage2 = pd.read_csv("../data/interim/youtube_cleaned_stage2.csv")
print(f"Loaded {len(df_stage2)} rows from checkpoint")
```

```python
from langdetect import detect_langs, LangDetectException

def detect_language_confidence(text):
    try:
        results = detect_langs(str(text))
        top = results[0]
        return top.lang, top.prob
    except LangDetectException:
        return None, 0.0

# This takes a few minutes on 70k+ rows — that's expected, langdetect isn't fast
lang_results = df_stage2["text_clean"].apply(detect_language_confidence)
df_stage2["detected_lang"] = lang_results.apply(lambda x: x[0])
df_stage2["detected_lang_prob"] = lang_results.apply(lambda x: x[1])

print(df_stage2["detected_lang"].value_counts().head(10))
```

```python
for threshold in [0.90, 0.95, 0.99]:
    count = ((df_stage2["detected_lang"] == "en") & (df_stage2["detected_lang_prob"] >= threshold)).sum()
    print(f"Threshold {threshold}: {count} comments would be flagged as high-confidence English "
          f"({count / len(df_stage2):.1%})")
```

```python
high_conf_english = df_stage2[
    (df_stage2["detected_lang"] == "en") & (df_stage2["detected_lang_prob"] >= 0.99)
]
print(f"{len(high_conf_english)} rows at 0.99+ confidence")
high_conf_english.sample(min(15, len(high_conf_english)))[["text_clean", "detected_lang_prob"]]
```

```python
import pandas as pd

df_stage4 = pd.read_csv("../data/interim/youtube_codemixed_stage4.csv")
print(f"Loaded {len(df_stage4)} rows from checkpoint")
```

```python
import re
import string

EMOJI_PATTERN = re.compile(
    "["
    "\U0001F300-\U0001FAFF"
    "\U00002600-\U000027BF"
    "\U0001F1E6-\U0001F1FF"
    "]+", flags=re.UNICODE
)

def strip_emoji(text):
    return EMOJI_PATTERN.sub("", str(text))

def is_mostly_repeated_char_v2(text, threshold=0.6):
    text_no_emoji = strip_emoji(text).strip()
    if len(text_no_emoji) < 3:
        return False
    most_common = max(set(text_no_emoji), key=text_no_emoji.count)
    return text_no_emoji.count(most_common) / len(text_no_emoji) > threshold

# Stricter: full match, not prefix match, and only bare/near-bare instances
GENERIC_PATTERNS_V2 = [
    r"^first!*$",
    r"^who('?s| is) here!*$",
    r"^nice video!*$",
    r"^\d+(st|nd|rd|th) comment!*$",
    r"^love (you|u)( bro| bhai)?!*$",
    r"^subscribe!*$",
    r"^notification squad!*$",
]
generic_re_v2 = re.compile("|".join(GENERIC_PATTERNS_V2), re.IGNORECASE)

df_stage4["is_repeated_char_v2"] = df_stage4["text_clean"].apply(is_mostly_repeated_char_v2)
df_stage4["is_generic_v2"] = df_stage4["text_clean"].apply(lambda t: bool(generic_re_v2.fullmatch(str(t).strip())))

print(f"Mostly repeated chars (v2, emoji-excluded): {df_stage4['is_repeated_char_v2'].sum()}")
print(f"Generic templated (v2, full match only): {df_stage4['is_generic_v2'].sum()}")
```

```python
print("=== High emoji ratio sample ===")
display(df_stage4[df_stage4["emoji_ratio"] > 0.5].sample(min(15, (df_stage4["emoji_ratio"] > 0.5).sum()))[["text_clean"]])

print("\n=== Repeated-char sample ===")
display(df_stage4[df_stage4["is_repeated_char_v2"]].sample(min(15, df_stage4["is_repeated_char"].sum()))[["text_clean"]])

print("\n=== Generic templated sample ===")
display(df_stage4[df_stage4["is_generic"]].sample(min(15, df_stage4["is_generic"].sum()))[["text_clean"]])
```

```python
print("=== Repeated-char v2 sample ===")
flagged_repeated = df_stage4[df_stage4["is_repeated_char_v2"]]
display(flagged_repeated.sample(min(15, len(flagged_repeated)))[["text_clean"]])

print("\n=== Generic templated v2 sample ===")
flagged_generic = df_stage4[df_stage4["is_generic_v2"]]
display(flagged_generic.sample(min(15, len(flagged_generic)))[["text_clean"]])
```
